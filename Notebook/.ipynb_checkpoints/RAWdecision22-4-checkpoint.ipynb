## <span style="color:palevioletred;">Selection Strategy</span>

To minimize the impact of electrode bridging, recording interruptions, and other data quality issues, subjects were evaluated individually according to predefined criteria.

Only recordings meeting acceptable quality standards were included in subsequent analyses.
### Evaluation Criteria

- Number of detected bridge pairs based on posibility from NONE to APPLE on scale 0-4
- Presence of recording interruptions
- Number of bad channels
- Overall signal quality
- Availability of usable recording duration
