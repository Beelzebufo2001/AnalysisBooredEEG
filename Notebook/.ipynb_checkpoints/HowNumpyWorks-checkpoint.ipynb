<h2>Intro to storage</h2>

Numpy is library mainly for numerical computing, handling large arrays and matrices as efficiently as posible. That's why it for Python-be-Simple lizard like me looks like door opening both ways. It's significantly faster than lists or any other Python build-in objects as is it implemented in C language style storage where <b>actual values</b> are stored at <b>contiguous location</b> AND not as object reference.

<h3>Python</h3> 

Python as an overcafinated squirrel memory, stores normally an object and variables holds the pointers(references) to those object.

```Python3
x = 3
# x -> [Python int object: value = 3, type = int, metadata...]
```
That x object is not just a THE number, it has type stored, reference counter, memory, value... So even 3 becomes chunky object in memory. List is even worse, since it points to continuous block of pointers, pointing on objects of: type, memory overhead, value... somewhere else

<u> >>> They are not contiguous

<h3>Numpy using C</h3>

```Python3
import numpy as np
a = np.array([1,2,3], dtype=np.int32)

#[1][2][3] as a value value value in contiguous way in 4bytes as 32-bit integer
```
This is what contiguous memory means:

>Address: 1000 1004 1008
>
>Values:    1    2    3

Each element sits right next to the previous one. CPUs love this because of cache locality. The processor loads chunks of nearby memory at once.

<h3>Numpy has a Dtype</h3>

Every element in has the same type. And that for me explains why values can be stored directly in the array and not only as a chunky Python object. We avoid the chunkiness using something like C language storage system with only one type for everything so that we dont have to create Python object .
